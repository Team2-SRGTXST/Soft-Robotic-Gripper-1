```python
import os
import shutil

negative_folder = "/content/negative_images"
train_images = "/content/peach-1/train/images"
train_labels = "/content/peach-1/train/labels"

count = 0

for filename in os.listdir(negative_folder):
    if filename.lower().endswith((".jpg", ".jpeg", ".png")):

        # Copy negative image into YOLO training images
        source = os.path.join(negative_folder, filename)
        destination = os.path.join(train_images, filename)
        shutil.copy2(source, destination)

        # Create matching EMPTY YOLO label file
        base_name = os.path.splitext(filename)[0]
        label_path = os.path.join(train_labels, base_name + ".txt")

        with open(label_path, "w") as f:
            pass

        count += 1

print(f"Added {count} negative images to training set.")
```

```python
import os

image_dir = "/content/peach-1/train/images"
label_dir = "/content/peach-1/train/labels"

images = [
    f for f in os.listdir(image_dir)
    if f.lower().endswith((".jpg", ".jpeg", ".png"))
]

positive = 0
negative = 0
missing = []

for image in images:
    name = os.path.splitext(image)[0]
    label = os.path.join(label_dir, name + ".txt")

    if not os.path.exists(label):
        missing.append(image)

    elif os.path.getsize(label) == 0:
        negative += 1

    else:
        positive += 1

print("Total images:", len(images))
print("Positive peach images:", positive)
print("Negative images:", negative)
print("Images missing labels:", len(missing))

if missing:
    print("\nMissing labels:")
    print(missing[:20])
```

```python
import os
import random
import matplotlib.pyplot as plt
from PIL import Image

image_folder = "/content/peach-1/train/images"
label_folder = "/content/peach-1/train/labels"

positive_images = []

for filename in os.listdir(image_folder):
    if filename.lower().endswith((".jpg", ".jpeg", ".png")):
        name = os.path.splitext(filename)[0]
        label = os.path.join(label_folder, name + ".txt")

        if os.path.exists(label) and os.path.getsize(label) > 0:
            positive_images.append(filename)

selected = random.sample(positive_images, 30)

plt.figure(figsize=(15, 20))

for i, filename in enumerate(selected):
    img = Image.open(os.path.join(image_folder, filename))

    plt.subplot(6, 5, i + 1)
    plt.imshow(img)
    plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
from PIL import Image
import os

image = Image.open("/content/peach_collage.png")

output_folder = "/content/new_peaches"
os.makedirs(output_folder, exist_ok=True)

width, height = image.size

cols = 6
rows = 5

tile_width = width // cols
tile_height = height // rows

count = 1

for row in range(rows):
    for col in range(cols):

        left = col * tile_width
        top = row * tile_height
        right = (col + 1) * tile_width
        bottom = (row + 1) * tile_height

        tile = image.crop((left, top, right, bottom))

        filename = f"peach_new_{count:02d}.jpg"
        tile.save(
            os.path.join(output_folder, filename),
            quality=95
        )

        count += 1

print("Created:", count - 1, "individual peach images")
print("Saved in:", output_folder)
```
